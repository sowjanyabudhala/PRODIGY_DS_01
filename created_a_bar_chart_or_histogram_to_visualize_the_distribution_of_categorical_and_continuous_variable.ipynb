**bivarient analysis**
categorical vs numerical data

```python
sns.barplot(x='Gender',y='Age',color='purple',data=df)
```

**conclusion:**from this graph we conclude that females have higher average age when compared with males.

```python
sns.barplot(x='Label',y='Age',color='red',data=df)
```

conclusion:from this graph we conclude that Obese category has a higher average age than the Normalweight or Underweight categories, this could indicate that obesity tends to increase with age.

```python
sns.barplot(x='Gender',y='Height',color='yellow',data=df)
```

**conclusion:**from this graph we conclude that males are having higher height when compared to females

```python
sns.barplot(x='Gender',y='Weight',color='pink',data=df)
```

**conclusion:**from this graph we conclude that males have a significantly higher average weight compared to females.

```python
sns.barplot(x='Gender',y='BMI',color='red',data=df)
```

conclusion:from this graph we conclude that males have a higher average BMI than females, it could be due to different factors such as metabolic rates, muscle mass, lifestyle, or diet.
